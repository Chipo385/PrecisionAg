Crop yield prediction

```python
#Load package
library(randomForest)
```

```python
#Sample Data for moisture temperature and crop yield
soil_moisture <- c(23, 34, 29, 31, 28)#percentage
temperature <- c(15, 19, 21, 20, 18)#celsius
crop_yield <- c(69, 72, 75, 77, 73) #per hectare
```

```python
#Labels for irrigation and fertilizer
irrigation <- c(22, 25, 23, 19, 20)#water percentage
fertilizer <- c(6,5,8,9,8)#kilograms
```

```python
#DataFrame for our predictors
features <- data.frame(
    soil_moisture = soil_moisture,
    temperature = temperature,
    crop_yield = crop_yield
)
```

```python
#Train our models
irrigation_model <- randomForest(irrigation ~., data = features)
fertilizer_model <- randomForest(irrigation ~., data = features)
```

```python
new_data <- data.frame(
    soil_moisture = 27,
    temperature = 18,
    crop_yield = 76
)
```

```python
predicted_irrigation <- predict(irrigation_model, new_data = new_data)
predicted_fertilizer <- predict(fertilizer_model, new_data = new_data)
```

```python
#print out our predictions
cat(sprintf("predicted irrigation schedule: %.2f%%\n", predicted_irrigation))
cat(sprintf("predicted fertilizer schedule: %.2f%%\n", predicted_irrigation))
```
